```python
for i,description in enumerate(object_descriptions):
    path=f"outputs/llm_results/object_{i:03d}.json"
    if os.path.exists(path):
        continue
    _,part_spec=generate_part_decomposition(description)
    _,anchor_tree=generate_anchor_tree(description,part_spec)
    _,global_tree=generate_global_obb(description,part_spec)
    _,hierarchy_local_tree=generate_hierarchy_local(description,part_spec)
    result={}
    result["description"]=description
    result["part_spec"]=part_spec
    result["anchor_tree"]=anchor_tree
    result["global_tree"]=global_tree
    result["hierarchy_local_tree"]=hierarchy_local_tree
    with open(path,"w",encoding="utf-8") as f:
        json.dump(result,f,indent=4,ensure_ascii=False)
    print(f"第{i}个物体已完成")
```

```python
depth_results={}
depth_results["Anchor"]={}
depth_results["Global"]={}
depth_results["HierarchyLocal"]={}
for i in range(len(anchor_trees)):
    depth=compute_hierarchy_depth(anchor_trees[i])
    if depth<=1:
        group="Depth1"
    elif depth==2:
        group="Depth2"
    else:
        group="Depth3+"
    if group not in depth_results["Anchor"]:
        depth_results["Anchor"][group]=[]
        depth_results["Global"][group]=[]
        depth_results["HierarchyLocal"][group]=[]
    depth_results["Anchor"][group].append(evaluate_single_tree(anchor_trees[i],connection_lists[i]))
    depth_results["Global"][group].append(evaluate_single_tree(global_trees[i],connection_lists[i]))
    depth_results["HierarchyLocal"][group].append(evaluate_single_tree(hierarchy_local_trees[i],connection_lists[i]))
```

```python
def summarize_metric_groups(group_results,metric):
    summary={}
    for group,results in group_results.items():
        values=[r[metric] for r in results if r.get(metric) is not None]
        if len(values)>0:
            summary[group]={"mean":float(np.mean(values)),"std":float(np.std(values)),"n":len(values)}
    return summary
```

```python
import copy
perturbation_scales=[0.8,1.2,1.5]
perturbation_results=[]
for file in result_files:
    with open(file,"r",encoding="utf-8") as f:
        data=json.load(f)
    connections=infer_semantic_connections(data["part_spec"])
    if len(connections)==0:
        continue
    parent_name,child_name=connections[0]
    for scale_factor in perturbation_scales:
        anchor_tree=copy.deepcopy(data["anchor_tree"])
        global_tree=copy.deepcopy(data["global_tree"])
        local_tree=copy.deepcopy(data["hierarchy_local_tree"])
        prepare_hierarchy_local_tree(local_tree)
        scale_part_bbox(anchor_tree,child_name,scale_factor)
        scale_part_bbox(global_tree,child_name,scale_factor)
        scale_part_bbox(local_tree,child_name,scale_factor)
        compute_anchor_tree_world_transform(anchor_tree)
        add_global_transform(global_tree)
        compute_hierarchy_local_world_transform(local_tree)
        normalize_tree_scale(anchor_tree)
        normalize_tree_scale(global_tree)
        normalize_tree_scale(local_tree)
        record={}
        record["scale"]=scale_factor
        record["anchor"]=metric_mean_contact_deviation(anchor_tree,connections)
        record["global"]=metric_mean_contact_deviation(global_tree,connections)
        record["local"]=metric_mean_contact_deviation(local_tree,connections)
        perturbation_results.append(record)
```

```python
perturbation_summary={}
for scale_factor in perturbation_scales:
    records=[x for x in perturbation_results if x["scale"]==scale_factor]
    perturbation_summary[scale_factor]={}
    for method in ["anchor","global","local"]:
        values=[x[method] for x in records if x[method] is not None]
        perturbation_summary[scale_factor][method]={"mean":float(np.mean(values)),"std":float(np.std(values)),"n":len(values)}
perturbation_summary
```

```python
anchor_vs_global_perturb={}
for scale_factor in perturbation_scales:
    anchor_values=np.array([x["anchor"] for x in perturbation_results if x["scale"]==scale_factor and x["anchor"] is not None and x["global"] is not None],dtype=float)
    global_values=np.array([x["global"] for x in perturbation_results if x["scale"]==scale_factor and x["anchor"] is not None and x["global"] is not None],dtype=float)
    diff=global_values-anchor_values
    anchor_vs_global_perturb[scale_factor]={}
    anchor_vs_global_perturb[scale_factor]["anchor_mean"]=float(np.mean(anchor_values))
    anchor_vs_global_perturb[scale_factor]["global_mean"]=float(np.mean(global_values))
    anchor_vs_global_perturb[scale_factor]["mean_improvement"]=float(np.mean(diff))
    anchor_vs_global_perturb[scale_factor]["p_value"]=float(wilcoxon(diff).pvalue)
anchor_vs_global_perturb
```

```python
anchor_vs_local_perturb={}
for scale_factor in perturbation_scales:
    anchor_values=np.array([x["anchor"] for x in perturbation_results if x["scale"]==scale_factor and x["anchor"] is not None and x["local"] is not None],dtype=float)
    local_values=np.array([x["local"] for x in perturbation_results if x["scale"]==scale_factor and x["anchor"] is not None and x["local"] is not None],dtype=float)
    diff=local_values-anchor_values
    anchor_vs_local_perturb[scale_factor]={}
    anchor_vs_local_perturb[scale_factor]["anchor_mean"]=float(np.mean(anchor_values))
    anchor_vs_local_perturb[scale_factor]["local_mean"]=float(np.mean(local_values))
    anchor_vs_local_perturb[scale_factor]["mean_improvement"]=float(np.mean(diff))
    anchor_vs_local_perturb[scale_factor]["p_value"]=float(wilcoxon(diff).pvalue)
anchor_vs_local_perturb
```

```python
import copy
random_seeds=[11,22,33,44,55]
random_anchor_results=[]
for i,file in enumerate(result_files):
    with open(file,"r",encoding="utf-8") as f:
        data=json.load(f)
    connections=infer_semantic_connections(data["part_spec"])
    for seed in random_seeds:
        tree=copy.deepcopy(data["anchor_tree"])
        randomize_tree_anchors(tree,seed)
        compute_anchor_tree_world_transform(tree)
        normalize_tree_scale(tree)
        result=evaluate_single_tree(tree,connections)
        random_anchor_results.append({"object_index":i,"seed":seed,"result":result})
```

```python
random_anchor_summary={}
metrics=["part_volume_sum","bbox_penetration_ratio","parent_child_contact_error","spatial_relation_accuracy","hierarchy_attachment_error","hierarchy_edge_success_rate","relative_attachment_consistency"]
for metric in metrics:
    values=[x["result"][metric] for x in random_anchor_results if x["result"].get(metric) is not None]
    random_anchor_summary[metric]={"mean":float(np.mean(values)),"std":float(np.std(values))}
random_anchor_summary
```

```python
random_object_results=[]
for i in range(len(result_files)):
    object_records=[x["result"] for x in random_anchor_results if x["object_index"]==i]
    result={}
    for metric in metrics:
        values=[x[metric] for x in object_records if x.get(metric) is not None]
        result[metric]=float(np.mean(values)) if len(values)>0 else None
    random_object_results.append(result)
```
